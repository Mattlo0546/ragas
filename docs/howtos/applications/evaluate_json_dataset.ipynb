# Evaluating a JSON dataset with Ragas

This notebook demonstrates how to load a JSON file containing question-answer pairs and evaluate it using Ragas metrics.

```
# install dependencies
!pip install -q ragas datasets
```

```
import json
from ragas import EvaluationDataset, evaluate
from ragas.metrics import AnswerRelevancy, Faithfulness

# path to your dataset
dataset_path = '../_static/simple_eval_dataset.json'

with open(dataset_path, 'r') as f:
    data = json.load(f)

# convert list of dicts to EvaluationDataset
eval_dataset = EvaluationDataset.from_list(data)
```

```
metrics = [AnswerRelevancy(), Faithfulness()]
result = evaluate(eval_dataset, metrics=metrics)
result
```
